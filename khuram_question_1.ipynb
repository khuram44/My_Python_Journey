list example

```python
my_list = [10, 20, 30, 40]


# Creating a list of numbers
numbers = [10, 20, 30, 40]

# Accessing an element
print(numbers[0])

# Changing an element
numbers[1] = 25

print(numbers)
```

Tuple

```python
my_tuple = (10, 20, 30, 40)
# Creating a tuple
numbers = (10, 20, 30, 40)
# Accessing an element
print(numbers[0])
# Displaying the complete tuple
print(numbers)
```

Set

```python
my_set = {10, 20, 30, 40}
# Creating a set with duplicate values
numbers = {10, 20, 20, 30, 40}
# Displaying the set
print(numbers)
# Adding a new element
numbers.add(50)
print(numbers)
```

Dictionary

```python
my_dict = {"name": "Ali", "age": 20}
# Creating a dictionary
student = {
    "name": "Ali",
    "age": 20,
    "course": "Python"
}
# Accessing a value using its key
print(student["name"])
# Updating a value
student["age"] = 21
print(student)
```
